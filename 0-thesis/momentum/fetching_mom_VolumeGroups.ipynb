```python
import os
import pandas as pd
import sqlalchemy as db
from dotenv import load_dotenv
```

```python
load_dotenv()

# config and credentials
server = os.getenv('DB_SERVER')
database = os.getenv('DB_NAME')
username = os.getenv('DB_USER')
password = os.getenv('DB_PASSWORD')

# connection
engine = db.create_engine(
    'mssql://{}:{}@{}/{}?driver=ODBC+Driver+18+for+SQL+Server'.format(
        username, password, server, database
    )
)

# establish connection
connection = engine.connect()
```

### Volume Groups

```python
volume_groups = [(1, 250000), (250000, 500000), (500000, 1000000), (1000000, 5000000)]

for idx, group in enumerate(volume_groups):

    query = '''
        DECLARE @RC int
        EXECUTE @RC = [dbo].[sp_mom_VolumeGroups] 
            {}, -- RangeStart
            {} -- RangeEnd
    '''.format(group[0], group[1])

    # read sql
    df = pd.read_sql_query(query, connection)
    df.to_csv('data/mom-VolumeGroups-{}.csv'.format(idx), index=False)
```
